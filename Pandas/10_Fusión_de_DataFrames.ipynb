# Fusión de DataFrames con Pandas: merge, concat y join

```python
import pandas as pd

df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value1': [1,2,3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value2': [4,5,6]
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
```

## Uso de `merge()`

```python
inner_join = pd.merge(df1, df2, on='key', how='inner')
print(inner_join)
```

```python
outer_merged = pd.merge(df1, df2, on='key', how='outer')
print(outer_merged)
```

```python
left_merged = pd.merge(df1, df2, on='key', how='left')
print(left_merged)
```

## Uso de `concat()`

```python
df3 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
})

df4 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
}) 
```

```python
vertical_concat = pd.concat([df3, df4])
print(vertical_concat)
```

```python
horizontal_concat = pd.concat([df3, df4], axis=1)
print(horizontal_concat)
```

## Uso de `join()`

```python
df5 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
    }, 
    index = ['K0', 'K1', 'K2']
) 

df6 = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['DO', 'D1', 'D2']
    }, 
    index = ['K0', 'K2', 'K3']
)
```

```python
joined = df5.join(df6, how='inner')
print(joined)
```
